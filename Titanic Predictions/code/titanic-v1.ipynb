### Data Overview

##### **PassengerId** : Id of the Passenger                      - Integer Value
##### **Survived**    : Survival of the Passenger                - Integer Value (0 or 1)
##### **Pclass**      : Ticket Class of the Passenger            - Integer Value (1 to 3)
##### **Name**        : Name of the Passenger                    - String
##### **Sex**         : Gender of the Passenger                  - String( male of female )
##### **Age**         : Age of the Passenger                     - Floating Number
##### **SibSp**       : No. of Siblings/Spouses of the Passenger - Integer Value
##### **Parch**       : No. Parents/Children of the Passenger    - Integer Value
##### **Ticket**      : Ticket of the Passeneger                 - String
##### **Fare**        : Fare paid by the Passenger               - Floating Number
##### **Cabin**       : Cabin assigned to the Passenger          - String Value
##### **Embarked**    : Station at which the Passenger boarded   - Char (Capital) (S, C, Q)

```python
dummy_full[['Title']] = dummy_full[['Name']].replace('(.*, )|(\\..*)', '', regex = True)
Title_tab = dummy_full.groupby(['Sex', 'Title']).size()
Title_tab.unstack(fill_value = 0)
```

```python
rare_title = ['Capt','Col','Don','Dona','Dr','Jonkheer','Lady','Major','Rev','Sir','the Countess']
     
#dummy_full['Title'][dummy_full['Title'] == 'Miss'] = 'Mrs'
dummy_full['Title'][dummy_full['Title'] == 'Mlle'] = 'Miss'
dummy_full['Title'][dummy_full['Title'] == 'Ms'] = 'Miss'
dummy_full['Title'][dummy_full['Title'] == 'Mme'] = 'Mrs'
dummy_full['Title'] = dummy_full['Title'].replace(
    dict.fromkeys(rare_title, 'rare_title')
)

Title_tab = dummy_full.groupby(['Sex', 'Title']).size()
Title_tab.unstack(fill_value = 0)
```

```python
def get_surname(data):
    name = data['Name']
    surname = ''
    for i in range(len(name)):
        if(name[i] == ','):
            break
        else:
            surname += name[i]
    return surname

surnames = []
for i in range(dummy_full.shape[0]):
    surnames.append(get_surname(dummy_full.iloc[i]))

sur = pd.DataFrame({
    'PassengerId2': dummy_full.PassengerId,
    'Surname' : surnames
})
# print(get_surname(dummy_full.iloc[0]))
#dummy_full.tail(6)
surnames[0]
dummy_full = pd.concat([dummy_full, sur] , axis = 1)
dummy_full = dummy_full.drop(['PassengerId2'], axis = 1)
dummy_full.head(4)
```
